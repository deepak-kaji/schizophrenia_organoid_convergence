```python
import os
import gc
import random

import numpy as np
import pandas as pd

import torch

import matplotlib.pyplot as plt
import seaborn as sns

import scanpy as sc
import scvi

%matplotlib inline

sc.settings.n_jobs = 25

random.seed(0)
np.random.seed(0)
sc.settings.seed = 0

torch.set_float32_matmul_precision("high")
```

```python
adata_hvg = sc.read_h5ad('/mnt/sdb/scz_meta_analysis_processed/anndata_objs/integrated_adata_hvg.h5ad')
```

```python
## Save and Loading ##

model_dir = os.path.join("/mnt/sdb/scz_meta_analysis_processed/models", "scvi_model")
model = scvi.model.SCVI.load(model_dir, adata=adata_hvg)

adata_hvg.layers["scvi_normalized"] = model.get_normalized_expression(library_size=10e4)

np.save("/mnt/sdb/scz_meta_analysis_processed/anndata_objs/scvi_normalized.npy", adata_hvg.layers["scvi_normalized"])
```
