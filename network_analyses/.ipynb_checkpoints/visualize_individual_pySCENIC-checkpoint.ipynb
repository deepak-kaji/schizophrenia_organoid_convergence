```python
# Convert all_regulons into an easy-to-read DataFrame

regulon_rows = []

for manuscript_name, regulons in all_regulons.items():

    for regulon in regulons:

        regulon_rows.append({
            "manuscript": manuscript_name,
            "regulon": regulon.name,
            "TF": regulon.transcription_factor,
            "n_targets": len(regulon.gene2weight),
            "targets": ", ".join(regulon.gene2weight.keys()),
            "score": regulon.score,
            "NES": regulon.nes,
            "context": ", ".join(regulon.context),
            "annotation": regulon.annotation
        })

regulon_df = pd.DataFrame(regulon_rows)

# Sort for easier inspection
regulon_df = regulon_df.sort_values(
    ["manuscript", "TF"]
).reset_index(drop=True)
```

```python
# do the regulons with governing TFs from different manuscripts have the same targets?

regulon_df["target_set"] = regulon_df["targets"].apply(lambda x: set(x.split(", ")))

tf_programs = (regulon_df.groupby(["manuscript", "TF"])["target_set"].agg(lambda sets: set().union(*sets)).reset_index())

tf_programs["n_targets"] = tf_programs["target_set"].str.len()
```

```python

def jaccard(a, b):
    return len(a & b) / len(a | b) if len(a | b) > 0 else np.nan


overlap_records = []

for tf, group in tf_programs.groupby("TF"):
    
    # Only TFs present in >=2 manuscripts
    if group["manuscript"].nunique() < 2:
        continue
    
    for (_, row1), (_, row2) in combinations(group.iterrows(), 2):
        
        targets1 = row1["target_set"]
        targets2 = row2["target_set"]
        shared = targets1 & targets2
        
        overlap_records.append({
            "TF": tf,
            "manuscript1": row1["manuscript"],
            "manuscript2": row2["manuscript"],
            "n_targets1": len(targets1),
            "n_targets2": len(targets2),
            "n_shared": len(shared),
            "jaccard": jaccard(targets1, targets2),
            "fraction_1_shared": len(shared) / len(targets1),
            "fraction_2_shared": len(shared) / len(targets2),
        })

tf_overlap_df = pd.DataFrame(overlap_records)
```

```python
# Calculate mean AUCell for Psychosis = 0 vs 1
regulon_means = (
    plot_long
    .groupby(["CellType", "Regulon", "Psychosis"])["AUCell"]
    .mean()
    .unstack("Psychosis")
)

# Difference: Psychosis (1) - No Psychosis (0)
regulon_means["difference"] = (
    regulon_means[1] - regulon_means[0]
)

# Rank separately within each cell type
top_regulons = {}

for celltype in plot_long["CellType"].unique():

    celltype_means = regulon_means.loc[celltype].copy()

    celltype_means["abs_difference"] = (
        celltype_means["difference"].abs()
    )

    top_regulons[celltype] = (
        celltype_means
        .nlargest(10, "abs_difference")
        .index
        .tolist()
    )

# Plot
celltypes = plot_long["CellType"].unique()

fig, axes = plt.subplots(
    len(celltypes),
    1,
    figsize=(14, 4 * len(celltypes)),
    squeeze=False
)

for ax, celltype in zip(axes.flat, celltypes):

    cell_df = plot_long[
        (plot_long["CellType"] == celltype) &
        (plot_long["Regulon"].isin(top_regulons[celltype]))
    ].copy()

    # Order by largest absolute Psychosis difference
    order = (
        regulon_means
        .loc[celltype]
        .assign(
            abs_difference=lambda x: x["difference"].abs()
        )
        .nlargest(10, "abs_difference")
        .index
    )

    sns.violinplot(
        data=cell_df,
        x="Regulon",
        y="AUCell",
        hue="Psychosis",
        order=order,
        split=True,
        inner="box",
        cut=0,
        ax=ax
    )

    ax.set_title(celltype)
    ax.set_xlabel("Regulon")
    ax.set_ylabel("Mean AUCell score")
    ax.tick_params(axis="x", rotation=45)
    ax.legend(title="Psychosis")

plt.tight_layout()
plt.show()
```

```python
auc_df_pb_idio['Manuscript'] = auc_df_pb_idio['Manuscript_Psychosis'].str.split('_').str[0]
```

```python
# ============================================================
# Top 10 regulons per cell type
# with all 4 Manuscript × Psychosis groups shown together
# ============================================================

plot_df = auc_df_pb_idio[
    shared_idio_regulons.tolist() +
    ["Psychosis", "CellType", "Manuscript", "Manuscript_Psychosis"]
].copy()

plot_long = plot_df.melt(
    id_vars=[
        "Psychosis",
        "CellType",
        "Manuscript",
        "Manuscript_Psychosis"
    ],
    value_vars=shared_idio_regulons.tolist(),
    var_name="Regulon",
    value_name="AUCell"
)

# ------------------------------------------------------------
# 1. Calculate Psychosis effect separately within each
#    CellType × Manuscript × Regulon
# ------------------------------------------------------------

regulon_means = (
    plot_long
    .groupby(
        ["CellType", "Manuscript", "Regulon", "Psychosis"]
    )["AUCell"]
    .mean()
    .unstack("Psychosis")
)

# Psychosis (1) - No Psychosis (0)
regulon_means["difference"] = (
    regulon_means[1] - regulon_means[0]
)

# ------------------------------------------------------------
# 2. Select top 10 regulons per cell type
#    requiring the Psychosis effect to be in the SAME
#    direction in both manuscripts
# ------------------------------------------------------------

top_regulons = {}

for celltype in plot_long["CellType"].unique():

    celltype_means = regulon_means.loc[celltype].copy()

    # Direction of Psychosis effect in each manuscript
    direction = (
        np.sign(celltype_means["difference"])
        .unstack("Manuscript")
    )

    # Average magnitude of effect across manuscripts
    magnitude = (
        celltype_means["difference"]
        .abs()
        .groupby(level="Regulon")
        .mean()
    )

    # Same direction in both manuscripts
    consistent = (
        (direction > 0).all(axis=1) |
        (direction < 0).all(axis=1)
    )

    # Top 10 consistent regulons
    top_regulons[celltype] = (
        magnitude[consistent]
        .nlargest(5)
        .index
        .tolist()
    )

    print(f"\n{celltype}")
    print(top_regulons[celltype])


# ------------------------------------------------------------
# 3. Set order of the four groups
# ------------------------------------------------------------

print(
    plot_long["Manuscript_Psychosis"]
    .unique()
)

group_order = [
    "Notaras_0",
    "Notaras_1",
    "Sawada_0",
    "Sawada_1"
]


# ------------------------------------------------------------
# 4. Plot
# ------------------------------------------------------------

celltypes = plot_long["CellType"].unique()

fig, axes = plt.subplots(
    len(celltypes),
    1,
    figsize=(16, 4 * len(celltypes)),
    squeeze=False
)

for ax, celltype in zip(axes.flat, celltypes):

    cell_df = plot_long[
        (plot_long["CellType"] == celltype) &
        (plot_long["Regulon"].isin(top_regulons[celltype]))
    ].copy()

    # Order regulons by average Psychosis effect
    order = (
        regulon_means
        .loc[celltype]
        .loc[
            regulon_means.loc[celltype]
            .index
            .get_level_values("Regulon")
            .isin(top_regulons[celltype])
        ]
        ["difference"]
        .groupby("Regulon")
        .mean()
        .sort_values()
        .index
        .tolist()
    )

    sns.violinplot(
        data=cell_df,
        x="Regulon",
        y="AUCell",
        hue="Manuscript_Psychosis",
        order=order,
        hue_order=group_order,
        inner="box",
        cut=0,
        ax=ax
    )

    ax.set_title(celltype, fontsize=13)
    ax.set_xlabel("Regulon")
    ax.set_ylabel("Mean AUCell score")
    ax.tick_params(axis="x", rotation=45)

    ax.legend(
        title="Manuscript / Psychosis",
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Center AUCell to manuscript-specific non-psychosis baseline
#
# For each CellType × Manuscript × Regulon:
#
#   Adjusted AUCell =
#       AUCell - mean(AUCell in Psychosis=0)
#
# This makes:
#   Notaras 0  -> centered around 0
#   Notaras 1  -> change relative to Notaras baseline
#   Sawada  0  -> centered around 0
#   Sawada  1  -> change relative to Sawada baseline
# ============================================================

plot_df = auc_df_pb_idio[
    shared_idio_regulons.tolist() +
    ["Psychosis", "CellType", "Manuscript", "Manuscript_Psychosis"]
].copy()

plot_long = plot_df.melt(
    id_vars=[
        "Psychosis",
        "CellType",
        "Manuscript",
        "Manuscript_Psychosis"
    ],
    value_vars=shared_idio_regulons.tolist(),
    var_name="Regulon",
    value_name="AUCell"
)

# ------------------------------------------------------------
# 1. Calculate the mean AUCell in the NON-PSYCHOSIS group
#    for each CellType × Manuscript × Regulon
# ------------------------------------------------------------

baseline = (
    plot_long[plot_long["Psychosis"] == 0]
    .groupby(
        ["CellType", "Manuscript", "Regulon"]
    )["AUCell"]
    .mean()
    .rename("Baseline")
)

# ------------------------------------------------------------
# 2. Subtract manuscript-specific non-psychosis baseline
# ------------------------------------------------------------

plot_long = plot_long.join(
    baseline,
    on=["CellType", "Manuscript", "Regulon"]
)

plot_long["AUCell_adjusted"] = (
    plot_long["AUCell"] - plot_long["Baseline"]
)

# ------------------------------------------------------------
# 3. Calculate the Psychosis effect within each manuscript
#    using the CENTERED values
# ------------------------------------------------------------

regulon_means = (
    plot_long
    .groupby(
        ["CellType", "Manuscript", "Regulon", "Psychosis"]
    )["AUCell_adjusted"]
    .mean()
    .unstack("Psychosis")
)

# Psychosis effect:
# positive = higher in psychosis
# negative = lower in psychosis

regulon_means["difference"] = (
    regulon_means[1] - regulon_means[0]
)

# ------------------------------------------------------------
# 4. Select top 10 regulons per cell type
#    requiring SAME direction in Notaras and Sawada
# ------------------------------------------------------------

top_regulons = {}

for celltype in plot_long["CellType"].unique():

    celltype_means = regulon_means.loc[celltype].copy()

    # Direction of Psychosis effect in each manuscript
    direction = (
        np.sign(celltype_means["difference"])
        .unstack("Manuscript")
    )

    # Average magnitude across manuscripts
    magnitude = (
        celltype_means["difference"]
        .abs()
        .groupby(level="Regulon")
        .mean()
    )

    # Same direction in both manuscripts
    consistent = (
        (direction > 0).all(axis=1) |
        (direction < 0).all(axis=1)
    )

    # Top 10 consistent regulons
    top_regulons[celltype] = (
        magnitude[consistent]
        .nlargest(5)
        .index
        .tolist()
    )

    print(f"\n{celltype}")
    print(top_regulons[celltype])


# ------------------------------------------------------------
# 5. Check your four group names
# ------------------------------------------------------------

print(
    "\nManuscript_Psychosis groups:"
)

print(
    plot_long["Manuscript_Psychosis"]
    .unique()
)

# Change these if your actual labels differ
group_order = [
    "Notaras_0",
    "Notaras_1",
    "Sawada_0",
    "Sawada_1"
]


# ------------------------------------------------------------
# 6. Plot centered values
# ------------------------------------------------------------

celltypes = plot_long["CellType"].unique()

fig, axes = plt.subplots(
    len(celltypes),
    1,
    figsize=(16, 4 * len(celltypes)),
    squeeze=False
)

for ax, celltype in zip(axes.flat, celltypes):

    cell_df = plot_long[
        (plot_long["CellType"] == celltype) &
        (plot_long["Regulon"].isin(top_regulons[celltype]))
    ].copy()

    # Order regulons by average Psychosis effect
    order = (
        regulon_means
        .loc[celltype]
        .loc[
            regulon_means.loc[celltype]
            .index
            .get_level_values("Regulon")
            .isin(top_regulons[celltype])
        ]
        ["difference"]
        .groupby("Regulon")
        .mean()
        .sort_values()
        .index
        .tolist()
    )

    sns.violinplot(
        data=cell_df,
        x="Regulon",
        y="AUCell_adjusted",
        hue="Manuscript_Psychosis",
        order=order,
        hue_order=group_order,
        inner="box",
        cut=0,
        ax=ax
    )

    # Zero line = manuscript-specific non-psychosis baseline
    ax.axhline(
        0,
        color="black",
        linestyle="--",
        linewidth=1,
        alpha=0.7
    )

    ax.set_title(
        celltype,
        fontsize=13
    )

    ax.set_xlabel("Regulon")
    ax.set_ylabel(
        "AUCell relative to manuscript-specific\n"
        "non-psychosis mean"
    )

    ax.tick_params(
        axis="x",
        rotation=45
    )

    ax.legend(
        title="Manuscript / Psychosis",
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

plt.tight_layout()
plt.show()
```

```python
auc_df_pb_nrxn1['Manuscript'] = auc_df_pb_nrxn1['Manuscript_Psychosis'].str.split('_').str[0]
```

```python
# ============================================================
# Center AUCell to manuscript-specific non-psychosis baseline
#
# For each CellType × Manuscript × Regulon:
#
#   Adjusted AUCell =
#       AUCell - mean(AUCell in Psychosis=0)
#
# This makes:
#   Sebastian 0  -> centered around 0
#   Sebastian 1  -> change relative to Sebastian baseline
#   Fernando  0  -> centered around 0
#   Fernando  1  -> change relative to Fernando baseline
# ============================================================

plot_df = auc_df_pb_nrxn1[
    shared_nrxn1_regulons.tolist() +
    ["Psychosis", "CellType", "Manuscript", "Manuscript_Psychosis"]
].copy()

plot_long = plot_df.melt(
    id_vars=[
        "Psychosis",
        "CellType",
        "Manuscript",
        "Manuscript_Psychosis"
    ],
    value_vars=shared_nrxn1_regulons.tolist(),
    var_name="Regulon",
    value_name="AUCell"
)

# ------------------------------------------------------------
# 1. Calculate the mean AUCell in the NON-PSYCHOSIS group
#    for each CellType × Manuscript × Regulon
# ------------------------------------------------------------

baseline = (
    plot_long[plot_long["Psychosis"] == 0]
    .groupby(
        ["CellType", "Manuscript", "Regulon"]
    )["AUCell"]
    .mean()
    .rename("Baseline")
)

# ------------------------------------------------------------
# 2. Subtract manuscript-specific non-psychosis baseline
# ------------------------------------------------------------

plot_long = plot_long.join(
    baseline,
    on=["CellType", "Manuscript", "Regulon"]
)

plot_long["AUCell_adjusted"] = (
    plot_long["AUCell"] - plot_long["Baseline"]
)

# ------------------------------------------------------------
# 3. Calculate the Psychosis effect within each manuscript
#    using the CENTERED values
# ------------------------------------------------------------

regulon_means = (
    plot_long
    .groupby(
        ["CellType", "Manuscript", "Regulon", "Psychosis"]
    )["AUCell_adjusted"]
    .mean()
    .unstack("Psychosis")
)

# Psychosis effect:
# positive = higher in psychosis
# negative = lower in psychosis

regulon_means["difference"] = (
    regulon_means[1] - regulon_means[0]
)

# ------------------------------------------------------------
# 4. Select top 10 regulons per cell type
#    requiring SAME direction in Notaras and Sawada
# ------------------------------------------------------------

top_regulons = {}

for celltype in plot_long["CellType"].unique():

    celltype_means = regulon_means.loc[celltype].copy()

    # Direction of Psychosis effect in each manuscript
    direction = (
        np.sign(celltype_means["difference"])
        .unstack("Manuscript")
    )

    # Average magnitude across manuscripts
    magnitude = (
        celltype_means["difference"]
        .abs()
        .groupby(level="Regulon")
        .mean()
    )

    # Same direction in both manuscripts
    consistent = (
        (direction > 0).all(axis=1) |
        (direction < 0).all(axis=1)
    )

    # Top 10 consistent regulons
    top_regulons[celltype] = (
        magnitude[consistent]
        .nlargest(5)
        .index
        .tolist()
    )

    print(f"\n{celltype}")
    print(top_regulons[celltype])


# ------------------------------------------------------------
# 5. Check your four group names
# ------------------------------------------------------------

print(
    "\nManuscript_Psychosis groups:"
)

print(
    plot_long["Manuscript_Psychosis"]
    .unique()
)

# Change these if your actual labels differ
group_order = [
    "Sebastian_0",
    "Sebastian_1",
    "Fernando_0",
    "Fernando_1"
]


# ------------------------------------------------------------
# 6. Plot centered values
# ------------------------------------------------------------

celltypes = plot_long["CellType"].unique()

fig, axes = plt.subplots(
    len(celltypes),
    1,
    figsize=(16, 4 * len(celltypes)),
    squeeze=False
)

for ax, celltype in zip(axes.flat, celltypes):

    cell_df = plot_long[
        (plot_long["CellType"] == celltype) &
        (plot_long["Regulon"].isin(top_regulons[celltype]))
    ].copy()

    # Order regulons by average Psychosis effect
    order = (
        regulon_means
        .loc[celltype]
        .loc[
            regulon_means.loc[celltype]
            .index
            .get_level_values("Regulon")
            .isin(top_regulons[celltype])
        ]
        ["difference"]
        .groupby("Regulon")
        .mean()
        .sort_values()
        .index
        .tolist()
    )

    sns.violinplot(
        data=cell_df,
        x="Regulon",
        y="AUCell_adjusted",
        hue="Manuscript_Psychosis",
        order=order,
        hue_order=group_order,
        inner="box",
        cut=0,
        ax=ax
    )

    # Zero line = manuscript-specific non-psychosis baseline
    ax.axhline(
        0,
        color="black",
        linestyle="--",
        linewidth=1,
        alpha=0.7
    )

    ax.set_title(
        celltype,
        fontsize=13
    )

    ax.set_xlabel("Regulon")
    ax.set_ylabel(
        "AUCell relative to manuscript-specific\n"
        "non-psychosis mean"
    )

    ax.tick_params(
        axis="x",
        rotation=45
    )

    ax.legend(
        title="Manuscript / Psychosis",
        bbox_to_anchor=(1.02, 1),
        loc="upper left"
    )

plt.tight_layout()
plt.show()
```
