```python
## Import Organoid Data ##

scz_adata = sc.read_h5ad('/mnt/sdb/scz_meta_analysis_processed/anndata_objs/integrated_adata_annotated_w_celltypist_scanvi.h5ad')

# remove stressed glia and doublets
scz_adata = scz_adata[(~scz_adata.obs['pred_dbl'])].copy()

# define pseudobulk column 
scz_adata.obs['Donor_Sample'] = scz_adata.obs['Donor'].astype('str') + '_' + scz_adata.obs['Sample Name'].astype('str')
scz_adata.obs['Donor_Sample_Celltype'] = scz_adata.obs['Donor_Sample'] + '_' + scz_adata.obs['CellType'].astype('str')
gc.collect()
```

```python
# ei branch probs
branch_probs_ei = pd.read_csv('/mnt/sdb/scz_meta_analysis_processed/trajectory_analyses/ei_branch_probs.csv', index_col=0)
branch_probs_di = pd.read_csv('/mnt/sdb/scz_meta_analysis_processed/trajectory_analyses/direct_indirect_branch_probs.csv', index_col=0)
branch_probs_di['Broad_Genotype'] = scz_adata.obs['Broad_Genotype']

## pickle load gene trends
in_path = "/mnt/sdb/scz_meta_analysis_processed/trajectory_analyses/ei_gene_trends.pkl"

with open(in_path, "rb") as f:
    gene_trends = pickle.load(f)
```

```python
## Create Adata Objects ##

ei_adata = scz_adata[branch_probs_ei.index].copy()
di_adata = scz_adata[branch_probs_di.index].copy()

# add constants for grouping if necessary
ei_adata.obs['constant'] = 1
di_adata.obs['constant'] = 1

# merge on palantir series cols 
di_adata.obs = di_adata.obs.merge(branch_probs_di[['Direct', 'Indirect', 'palantir_pseudotime', 'Path',
                                                'palantir_entropy', 'assignment']], 
                                  left_index=True, right_index=True, how='left')

di_adata.obs['Path'] = di_adata.obs['Path'].astype('str')

ei_adata.obs = ei_adata.obs.merge(branch_probs_ei[['Excitatory', 'Inhibitory', 'palantir_pseudotime', 'Path',
                                                'palantir_entropy', 'assignment']], 
                                  left_index=True, right_index=True, how='left')

ei_adata.obs['Path'] = ei_adata.obs['Path'].astype('str')

# add broad genotype path
ei_adata.obs['Broad_Genotype_Path'] = ei_adata.obs['Broad_Genotype'].astype('str') + '_' + ei_adata.obs['Path'].astype('str')
di_adata.obs['Broad_Genotype_Path'] = di_adata.obs['Broad_Genotype'].astype('str') + '_' + di_adata.obs['Path'].astype('str')

## calculate neighbors 

sc.pp.neighbors(ei_adata, n_neighbors=15, use_rep='X_scVI')
sc.pp.neighbors(di_adata, n_neighbors=15, use_rep='X_scVI')
```

```python
## Recalculate linear pseudotime for DI anndata ##

# 1. Define mask
mask = ((di_adata.obs['subclass_annotations'] == 'Radial Glia') & (di_adata.obs['Day'] == 22))

# 2. Get indices in full AnnData
subset_idx = np.where(mask)[0]

# 3. Pull scVI latent space
X = di_adata.obsm['X_scVI'][subset_idx]

# 4. Compute centroid in scVI space
centroid = X.mean(axis=0)

# 5. Find closest cell
dists = pairwise_distances(X, centroid.reshape(1, -1)).ravel()
closest_local_idx = np.argmin(dists)

# 6. Map back to original index
iroot = subset_idx[closest_local_idx]

# 7. Set root
di_adata.uns['iroot'] = int(iroot)

## Calculate linear pseudotime
sc.tl.diffmap(di_adata)
sc.tl.dpt(di_adata)
```

```python
## Direct Indirect Heatmap ##

# -----------------------------
# 1. DEFINE CUSTOM ORDER
# -----------------------------
custom_order = ["Control_25", "Control_30",  "Idiopathic_Schizophrenia_25",  "Idiopathic_Schizophrenia_30",
                "15q13.3del_25", "15q13.3del_30", "22q11.2del_25", "22q11.2del_30",
                "3q29del_25", "3q29del_30",  "NRXN1del_25", "NRXN1del_30"]

# -----------------------------
# 2. REORDER DATA FIRST
# -----------------------------
df = auc_mtx_grouped_di.copy()
df = df.reindex(custom_order)

# -----------------------------
# 3. SELECT TOP REGULONS
# -----------------------------
results = {}

for genotype in df.index:
    row_vals = df.loc[genotype]
    other_max = df.drop(index=genotype).max(axis=0)
    score = row_vals - other_max
    top3 = score.sort_values(ascending=False).head(3).index
    results[genotype] = top3

top3_df = pd.DataFrame(results).T
top3_df.columns = ["Top1", "Top2", "Top3"]

selected_cols = pd.unique(top3_df.values.flatten())

selected_cols_di = selected_cols.copy()

heatmap_df = df[selected_cols]

# -----------------------------
# 4. COLUMN NORMALIZATION
# -----------------------------
heatmap_norm = (heatmap_df - heatmap_df.mean(axis=0)) / heatmap_df.std(axis=0)

# -----------------------------
# 5. ALIGN RSS
# -----------------------------
rss_sub = rss_di.loc[heatmap_norm.index, heatmap_norm.columns]

rss_scaled = (rss_sub - rss_sub.min().min()) / (rss_sub.max().max() - rss_sub.min().min())

sizes = (rss_scaled * 150)

# -----------------------------
# 6. PLOT
# -----------------------------
plt.figure(figsize=(6, 6))
ax = sns.heatmap(heatmap_norm, cmap=red_cmap, center=0, cbar=True)

for i in range(heatmap_norm.shape[0]):
    for j in range(heatmap_norm.shape[1]):
        if not np.isnan(heatmap_norm.iloc[i, j]):
            ax.scatter(
                j + 0.5,
                i + 0.5,
                s=sizes.iloc[i, j],
                color="white",
                edgecolor="black",
                linewidth=0.5,
                alpha=0.8)

plt.title("Top 3 regulons per genotype (AUC z-score + RSS size)")
plt.tight_layout()
plt.show()
```

```python
def plot_gene_pseudotime_raw(adata, gene, pseudotime_key="palantir_pseudotime", groupby=None, 
                             n_bins=20, layer=None, use_raw=False, figsize=(8,4), ci="sem", legend=True, save=None):



    # --- expression extraction ---
    expr = (adata.raw[:, gene].X if use_raw else
        adata.layers[layer][:, adata.var_names.get_loc(gene)] if layer else
        adata[:, gene].X)

    expr = expr.toarray().flatten() if not isinstance(expr, np.ndarray) else expr.flatten()

    # --- dataframe ---
    df = pd.DataFrame({
        "pseudotime": adata.obs[pseudotime_key].values,
        "Expression": expr
    })

    if groupby:
        df[groupby] = adata.obs[groupby].values

    df["pt_bin"] = pd.cut(df["pseudotime"], bins=n_bins)

    # --- aggregation ---
    gb = [groupby, "pt_bin"] if groupby else ["pt_bin"]

    df_binned = (
        df.groupby(gb, observed=True)
          .agg(
              pseudotime=("pseudotime", "mean"),
              mean=("Expression", "mean"),
              std=("Expression", "std"),
              n=("Expression", "count")
          )
          .reset_index()
    )

    # --- CI ---
    df_binned["err"] = (
        df_binned["std"] / np.sqrt(df_binned["n"])
        if ci == "sem" else df_binned["std"]
    )

    # --- smoothing ---
    if groupby:
        df_binned["smooth"] = df_binned.groupby(groupby)["mean"].transform(
            lambda x: x.rolling(3, center=True, min_periods=1).mean()
        )
        df_binned["err_smooth"] = df_binned.groupby(groupby)["err"].transform(
            lambda x: x.rolling(3, center=True, min_periods=1).mean()
        )
    else:
        df_binned["smooth"] = df_binned["mean"].rolling(3, center=True, min_periods=1).mean()
        df_binned["err_smooth"] = df_binned["err"].rolling(3, center=True, min_periods=1).mean()

    # --- plot ---
    fig, ax = plt.subplots(figsize=figsize)

    if groupby:
        for key, sub in df_binned.groupby(groupby):
            ax.plot(sub["pseudotime"], sub["smooth"], label=key)
            ax.fill_between(
                sub["pseudotime"],
                sub["smooth"] - sub["err_smooth"],
                sub["smooth"] + sub["err_smooth"],
                alpha=0.2
            )

        if legend:
            ax.legend(title=groupby, bbox_to_anchor=(1.05, 1), loc="upper left")
        else:
            if ax.get_legend() is not None:
                ax.get_legend().remove()

    else:
        ax.plot(df_binned["pseudotime"], df_binned["smooth"])
        ax.fill_between(
            df_binned["pseudotime"],
            df_binned["smooth"] - df_binned["err_smooth"],
            df_binned["smooth"] + df_binned["err_smooth"],
            alpha=0.2
        )

    # --- styling ---
    sns.despine()
    ax.set_title(f"{gene} expression over pseudotime")
    ax.set_ylabel("Smoothed expression")
    ax.set_xlabel("Pseudotime")

    # --- SAVE MODE (override aesthetics) ---
    if save:
        ax.set_title("")
        ax.set_ylabel("")
        ax.set_xlabel("")
        if ax.get_legend() is not None:
            ax.get_legend().remove()

        path = f"/home/deepak/datasets/scz_meta_analysis/panels/{save}.svg"
        plt.savefig(path, dpi=300, bbox_inches="tight")

    plt.show()
```

```python
def plot_gene_pseudotime_raw(
    adata,
    gene,
    pseudotime_key="palantir_pseudotime",
    groupby=None,
    n_bins=20,
    layer=None,
    use_raw=False,
    figsize=(8,4),
    ci="sem",
    legend=True,
    colors=None,  
    save=None):

    # --- expression extraction ---
    expr = (
        adata.raw[:, gene].X if use_raw else
        adata.layers[layer][:, adata.var_names.get_loc(gene)] if layer else
        adata[:, gene].X
    )

    expr = (
        expr.toarray().flatten()
        if not isinstance(expr, np.ndarray)
        else expr.flatten()
    )

    # --- dataframe ---
    df = pd.DataFrame({
        "pseudotime": adata.obs[pseudotime_key].values,
        "Expression": expr
    })

    if groupby:
        df[groupby] = adata.obs[groupby].values

    df["pt_bin"] = pd.cut(df["pseudotime"], bins=n_bins)

    # --- aggregation ---
    gb = [groupby, "pt_bin"] if groupby else ["pt_bin"]

    df_binned = (
        df.groupby(gb, observed=True)
          .agg(
              pseudotime=("pseudotime", "mean"),
              mean=("Expression", "mean"),
              std=("Expression", "std"),
              n=("Expression", "count")
          )
          .reset_index()
    )

    # --- CI ---
    df_binned["err"] = (
        df_binned["std"] / np.sqrt(df_binned["n"])
        if ci == "sem" else df_binned["std"]
    )

    # --- smoothing ---
    if groupby:
        df_binned["smooth"] = (
            df_binned.groupby(groupby)["mean"]
            .transform(
                lambda x: x.rolling(
                    3,
                    center=True,
                    min_periods=1
                ).mean()
            )
        )

        df_binned["err_smooth"] = (
            df_binned.groupby(groupby)["err"]
            .transform(
                lambda x: x.rolling(
                    3,
                    center=True,
                    min_periods=1
                ).mean()
            )
        )

    else:
        df_binned["smooth"] = (
            df_binned["mean"]
            .rolling(3, center=True, min_periods=1)
            .mean()
        )

        df_binned["err_smooth"] = (
            df_binned["err"]
            .rolling(3, center=True, min_periods=1)
            .mean()
        )

    # --- plot ---
    fig, ax = plt.subplots(figsize=figsize)

    if groupby:

        for i, (key, sub) in enumerate(df_binned.groupby(groupby)):

            # choose color
            if colors is not None:
                color = colors.get(key, None)
            else:
                color = None

            ax.plot(
                sub["pseudotime"],
                sub["smooth"],
                label=key,
                color=color
            )

            ax.fill_between(
                sub["pseudotime"],
                sub["smooth"] - sub["err_smooth"],
                sub["smooth"] + sub["err_smooth"],
                alpha=0.2,
                color=color
            )

        if legend:
            ax.legend(
                title=groupby,
                bbox_to_anchor=(1.05, 1),
                loc="upper left"
            )
        else:
            if ax.get_legend() is not None:
                ax.get_legend().remove()

    else:

        color = colors if isinstance(colors, str) else None

        ax.plot(
            df_binned["pseudotime"],
            df_binned["smooth"],
            color=color
        )

        ax.fill_between(
            df_binned["pseudotime"],
            df_binned["smooth"] - df_binned["err_smooth"],
            df_binned["smooth"] + df_binned["err_smooth"],
            alpha=0.2,
            color=color
        )

    # --- styling ---
    sns.despine()

    ax.set_title(f"{gene} expression over pseudotime")
    ax.set_ylabel("Smoothed expression")
    ax.set_xlabel("Pseudotime")

    # --- SAVE MODE ---
    if save:

        ax.set_title("")
        ax.set_ylabel("")
        ax.set_xlabel("")

        if ax.get_legend() is not None:
            ax.get_legend().remove()

        path = f"/home/deepak/datasets/scz_meta_analysis/panels/{save}.svg"

        plt.savefig(path, dpi=300, bbox_inches="tight")

    plt.show()
```

```python
# Regression Analysis ##

# Core columns (what you’re modeling)
# Column	Meaning
# estimate	log-odds effect (β)
# std.error	uncertainty of β
# statistic	z-score (β / SE)
# p.value	significance test on β
# conf.low, conf.high	CI on log-odds scale
# Converted columns
# Column	Meaning
# odds_ratio	exp(estimate)
# conf.low.or	exp(conf.low)
# conf.high.or	exp(conf.high)
# Formula
# log-odds (β)	OR = exp(β)
# OR	log-odds = log(OR)

# Path2/Path1 so Excitatory positive, Inhibitory Negative
# Dierct/Indirect so coefficients signal Direct positive, Indirect Negative

di_coefs = pd.read_csv('/mnt/sdb/scz_meta_analysis_processed/trajectory_analyses/lmer_di_pseudobulk_coefficients.csv')
bd_coefs_df_di = di_coefs[di_coefs['term'].str.contains('Broad')]
bd_coefs_df_di['term'] = bd_coefs_df_di['term'].str.replace('Broad_Genotype', '', regex=False).str.replace(r'X(?=\d)', '', regex=True)
bd_coefs_df_di['neglogP'] = -np.log10(bd_coefs_df_di['p.value'])

ei_coefs = pd.read_csv('/mnt/sdb/scz_meta_analysis_processed/trajectory_analyses/lmer_ei_pseudobulk_coefficients.csv')
bd_coefs_df_ei = ei_coefs[ei_coefs['term'].str.contains('Broad')]
bd_coefs_df_ei['term'] = bd_coefs_df_ei['term'].str.replace('Broad_Genotype', '', regex=False).str.replace(r'X(?=\d)', '', regex=True)
bd_coefs_df_ei['neglogP'] = -np.log10(bd_coefs_df_ei['p.value'])
```

```python
## Excitatory Inhibitory Forest Plot ##

df = bd_coefs_df_ei.sort_values("estimate")

sns.set_theme(style="white")

fig, ax = plt.subplots(figsize=(5, 4))

# --- size encodes significance ---
sizes = 30 + 25 * np.log1p(df["neglogP"])

# --- base color = effect size ---
norm = plt.Normalize(df["estimate"].min(), df["estimate"].max())
base_colors = plt.cm.coolwarm(norm(df["estimate"]))

# --- override non-significant points to gray ---
colors = base_colors.copy()

ns_mask = df["p.value"] > 0.05
colors[ns_mask] = np.array([0.7, 0.7, 0.7, 1.0])  # light gray RGBA

# --- scatter ---
ax.scatter(df["estimate"],  df["term"], s=sizes, c=colors, edgecolor="white", linewidth=0.8, zorder=2)

# --- CI with caps ---
ax.errorbar(x=df["estimate"], y=df["term"], xerr=[df["estimate"] - df["conf.low"], df["conf.high"] - df["estimate"]],
            fmt="none", ecolor="black", elinewidth=1.4, capsize=6, alpha=0.8, zorder=1)

# --- zero line ---
ax.axvline(0, linestyle="--", color="black", linewidth=1)

# --- labels ---
ax.set_xlabel("Effect size (log scale)", fontsize=11)
ax.set_ylabel("")

sns.despine()
ax.tick_params(axis='both', labelsize=10)
plt.savefig("/home/deepak/datasets/scz_meta_analysis/panels/Fig_6_ei_forest_plot.svg", dpi=500, bbox_inches="tight")
plt.tight_layout()
plt.show()
```

```python
def plot_genes_pseudotime(
    adata,
    genes,
    pseudotime_key="palantir_pseudotime",
    groupby=None,
    n_bins=20,
    layer=None,
    use_raw=False,
    figsize=(6, 4),
    ci="sem",
    smooth_window=3,
    sharey=True,
    save=None,
):

    # -----------------------------
    # build long dataframe
    # -----------------------------
    dfs = []

    for gene in genes:

        expr = (
            adata.raw[:, gene].X if use_raw else
            adata.layers[layer][:, adata.var_names.get_loc(gene)] if layer else
            adata[:, gene].X
        )

        expr = (
            expr.toarray().flatten()
            if not isinstance(expr, np.ndarray)
            else expr.flatten()
        )

        tmp = pd.DataFrame({
            "pseudotime": adata.obs[pseudotime_key].values,
            "Expression": expr,
            "Gene": gene
        })

        if groupby:
            tmp[groupby] = adata.obs[groupby].values

        dfs.append(tmp)

    df = pd.concat(dfs, ignore_index=True)

    # -----------------------------
    # pseudotime bins
    # -----------------------------
    df["pt_bin"] = pd.cut(df["pseudotime"], bins=n_bins)

    # grouping structure
    gb = ["Gene", "pt_bin"]

    if groupby:
        gb = [groupby] + gb

    # -----------------------------
    # aggregate
    # -----------------------------
    df_binned = (
        df.groupby(gb, observed=True)
        .agg(
            pseudotime=("pseudotime", "mean"),
            mean=("Expression", "mean"),
            std=("Expression", "std"),
            n=("Expression", "count")
        )
        .reset_index()
    )

    # -----------------------------
    # error bars
    # -----------------------------
    df_binned["err"] = (
        df_binned["std"] / np.sqrt(df_binned["n"])
        if ci == "sem"
        else df_binned["std"]
    )

    # -----------------------------
    # smoothing
    # -----------------------------
    smooth_group = ["Gene"]

    if groupby:
        smooth_group = [groupby, "Gene"]

    df_binned["smooth"] = (
        df_binned.groupby(smooth_group)["mean"]
        .transform(
            lambda x: x.rolling(
                smooth_window,
                center=True,
                min_periods=1
            ).mean()
        )
    )

    df_binned["err_smooth"] = (
        df_binned.groupby(smooth_group)["err"]
        .transform(
            lambda x: x.rolling(
                smooth_window,
                center=True,
                min_periods=1
            ).mean()
        )
    )

    # -----------------------------
    # plotting
    # -----------------------------
    if groupby:

        groups = df_binned[groupby].unique()
        n_groups = len(groups)

        fig, axes = plt.subplots(
            n_groups,
            1,
            figsize=(figsize[0], figsize[1] * n_groups),
            sharex=True,
            sharey=sharey
        )

        if n_groups == 1:
            axes = [axes]

        for ax, grp in zip(axes, groups):

            sub_grp = df_binned[df_binned[groupby] == grp]

            for gene in genes:

                sub = sub_grp[sub_grp["Gene"] == gene]

                ax.plot(
                    sub["pseudotime"],
                    sub["smooth"],
                    label=gene
                )

                ax.fill_between(
                    sub["pseudotime"],
                    sub["smooth"] - sub["err_smooth"],
                    sub["smooth"] + sub["err_smooth"],
                    alpha=0.15
                )

            ax.set_title(grp)
            ax.set_ylabel("Expression")

            sns.despine(ax=ax)

        axes[-1].set_xlabel("Pseudotime")

        axes[0].legend(
            bbox_to_anchor=(1.05, 1),
            loc="upper left"
        )

    else:

        fig, ax = plt.subplots(figsize=figsize)

        for gene in genes:

            sub = df_binned[df_binned["Gene"] == gene]

            ax.plot(
                sub["pseudotime"],
                sub["smooth"],
                label=gene
            )

            ax.fill_between(
                sub["pseudotime"],
                sub["smooth"] - sub["err_smooth"],
                sub["smooth"] + sub["err_smooth"],
                alpha=0.15
            )

        ax.legend()
        ax.set_xlabel("Pseudotime")
        ax.set_ylabel("Expression")

        sns.despine()

    # -----------------------------
    # save
    # -----------------------------
    if save:
        plt.savefig(
            save,
            dpi=300,
            bbox_inches="tight"
        )

    plt.show()
```

```python
plot_genes_pseudotime(ei_adata, genes=["VIM", "EOMES", "DLX6", "NEUROD6"], groupby='Path')
```
