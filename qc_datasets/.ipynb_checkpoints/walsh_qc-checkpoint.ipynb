```python
import glob
import gc
import random

import numpy as np
import pandas as pd

from scipy.stats import median_abs_deviation, percentileofscore

import matplotlib.pyplot as plt
import seaborn as sns

import scanpy as sc

import pegasus as pg
from pegasusio import UnimodalData, MultimodalData

from tqdm import tqdm

%matplotlib inline

sc.settings.n_jobs = 25

random.seed(0)
np.random.seed(0)
sc.settings.seed = 0
```

```python
## Import CellBendered Adata Object ##

walsh_adata = sc.read_h5ad('/mnt/sdb/scz_meta_analysis_processed/studies/walsh_studer/anndata_objs/walsh_cellbendered.h5ad')

walsh_adata.obs['Run'] = [i.split('_')[0] for i in walsh_adata.obs_names]

walsh_adata.obs['barcodes'] = walsh_adata.obs_names

# filter ambient rna
walsh_adata = walsh_adata[walsh_adata.obs['cell']].copy()
```

```python
## Cells After Ambient RNA Removal ##

walsh_adata.shape[0]
```

```python
## Merge SRA Metadata ##

walsh_sra_metadata = pd.read_csv('/mnt/sda/scz_meta_analysis/walsh_studer_neuron_2025/metadata/Walsh_SRA_RunTable.csv')

walsh_sra_metadata_essential = walsh_sra_metadata[['Run', 'BioProject', 'BioSample',
                                                   'cell_line', 'Center Name', 
                                                   'Experiment', 'genotype', 'Instrument',
                                                   'Library Name', 'ReleaseDate', 'create_date', 'Sample Name']]

walsh_adata.obs = walsh_adata.obs.merge(walsh_sra_metadata_essential, left_on='Run', right_on='Run', how='left')

walsh_adata.obs.index = walsh_adata.obs['barcodes']

# MEL-1 Line Idnetified at Male in Landmark Paper: https://www.nature.com/articles/nbt1318 Supplementary Table 6

walsh_adata.obs['Sex'] = 'Male'
```

```python
## Extra Metadata Obtained Through Personal Communication with Ryan Walsh ##

# What day was this sample dissociated?
sample_day_dict = {'GSM7979671':75, 'GSM7979672': 75, 'GSM7979673':60, 'GSM7979675':60, 'GSM7979674':60, 
                   'GSM7979676':60, 'GSM7979677':125, 'GSM7979679':125, 'GSM7979678':125, 'GSM7979680':125}

walsh_adata.obs['Day'] = walsh_adata.obs['Sample Name'].apply(lambda x: sample_day_dict[x])

# What structure was dissociated and sorted with flow cytometry for ventral cells?
sample_organoid_origin_dict = {'GSM7979671': 'Dorsal', 'GSM7979672': 'Ventral', 'GSM7979673': 'D-V Assembloid', 
                               'GSM7979675':'D-V Assembloid', 'GSM7979674':'D-V Assembloid', 'GSM7979676':'D-V Assembloid', 
                               'GSM7979677':'D-V Assembloid', 'GSM7979679':'D-V Assembloid', 'GSM7979678':'D-V Assembloid', 
                               'GSM7979680':'D-V Assembloid'}

walsh_adata.obs['Organoid_Origin'] = walsh_adata.obs['Sample Name'].apply(lambda x: sample_organoid_origin_dict[x])

# What cells were sorted for
walsh_adata.obs['Differentiation'] = 'Ventral'
```

```python
## Map NCBI Gene Names w/Prasanth's File ##

# load prasanth's annotation file
prasanth_annot = pd.read_csv('/home/deepak/datasets/annotations/all_genes_EnsemblInfo_104.csv')

# merge prasanth's gene info into the var dataframe
walsh_adata.var['ensembl_gene_id'] = walsh_adata.var.index

walsh_adata.var = walsh_adata.var.merge(prasanth_annot, left_on='ensembl_gene_id', right_on='ensembl_gene_id', how='left')
walsh_adata.var.chromosome_name = walsh_adata.var.chromosome_name.astype('str')

# fix gene name that is NaN
walsh_adata.var.loc[walsh_adata.var['hgnc_symbol'].isna(),'hgnc_symbol'] = walsh_adata.var.loc[walsh_adata.var['hgnc_symbol'].isna(),'ensembl_gene_id']

#reset the index as the hgnc_symbol
walsh_adata.var.index = walsh_adata.var['hgnc_symbol']
walsh_adata.var.index.name = None

#some ensembls map to same hgnc so we make them unique
walsh_adata.var_names_make_unique(join='#')

gc.collect()
```

```python
## Donor Metadata From Supplementary Table 2 of Paper ##

## no real description of the lines, all scRNASEq done in MEL-1 ##

walsh_adata.obs.rename({'cell_line': 'Donor'}, inplace=True, axis=1)

walsh_adata.obs['Sex'] = walsh_adata.obs['Sex'].astype('str')
```

```python
## Remove Duplicate Barcodes ##

print(walsh_adata.obs_names.duplicated(keep=False).any())
print(walsh_adata.var_names.duplicated(keep=False).any())

# None
```

```python
## Set Up QC Metrics ##

# filter on robust genes
# annotate_robust_genes_scanpy(integrated, percent_cells=0.05)
pg.identify_robust_genes(walsh_adata, percent_cells=0.05)
walsh_adata = walsh_adata[:,walsh_adata.var['robust']].copy()
gc.collect()

# mitochondrial genes, "MT-" for human, "Mt-" for mouse
walsh_adata.var["mt"] = walsh_adata.var_names.str.startswith("MT-")
# ribosomal genes
walsh_adata.var["ribo"] = walsh_adata.var_names.str.startswith(("RPS", "RPL"))
# hemoglobin genes
walsh_adata.var["hb"] = walsh_adata.var_names.str.contains("^HB[^(P)]")

# add protein_coding genes
walsh_adata.var['protein_coding'] = walsh_adata.var['gene_biotype'].apply(lambda x: True if 'protein_coding' == x else False)
walsh_adata.var['protein_coding'] = walsh_adata.var['protein_coding'].astype('bool')

# define mitocarta_genes
mitocarta_df = pd.read_csv('/home/deepak/datasets/annotations/Human.MitoCarta3.0.csv')
walsh_adata.var['mitocarta'] = [True if x in list(mitocarta_df['Symbol']) else False for x in walsh_adata.var.index]

# define robust_protein_coding genes (exclude ribosomal (RPL,RPS), mitochondrial, or mitocarta genes
walsh_adata.var['robust_protein_coding'] = walsh_adata.var['robust'] & walsh_adata.var['protein_coding']
walsh_adata.var.loc[walsh_adata.var.ribo, 'robust_protein_coding'] = False
walsh_adata.var.loc[walsh_adata.var.mt, 'robust_protein_coding'] = False
walsh_adata.var.loc[walsh_adata.var.mitocarta, 'robust_protein_coding'] = False

# define robust_protein_coding_autosome genes (exclude ribosomal (RPL,RPS), mitochondrial, or mitocarta genes
walsh_adata.var['autosome'] = walsh_adata.var.chromosome_name.isin([str(i) for i in range(1,23)])
walsh_adata.var['robust_protein_coding_autosome'] = walsh_adata.var['robust_protein_coding'] & walsh_adata.var['autosome'] 
```

```python
## Plot QC Metrics ##

# QC by counts #
# convert to pegasus unimodal object
walsh_adata = UnimodalData(walsh_adata)
pg.qc_metrics(walsh_adata, mito_prefix='MT-')

# Create a figure with 3 subplots

# convert back for plotting to scanpy anndata
walsh_adata = walsh_adata.to_anndata()
fig, axes = plt.subplots(2, 2, figsize=(10, 10))
axes = axes.flatten()
sc.pl.violin(walsh_adata, "n_genes", ax=axes[0], show=False, stripplot=False)
sc.pl.violin(walsh_adata, "n_counts", ax=axes[1], show=False, stripplot=False)
sc.pl.violin(walsh_adata, "percent_mito", ax=axes[2], show=False, stripplot=False)
fig.delaxes(axes[3])
plt.tight_layout()

sc.pl.scatter(walsh_adata, "n_counts", "n_genes", color="percent_mito")

print('The 99th percentile all sample genes/cell is: {0}'.format(np.percentile(walsh_adata.obs['n_genes'], 95)))

print('The 99th percentile all sample total counts/cell is: {0}'.format(np.percentile(walsh_adata.obs['n_counts'], 95)))

print('The 99th percentile all sample mitochondrial % /cell is: {0}'.format(np.percentile(walsh_adata.obs['percent_mito'], 95)))
```

```python
## Visualizating QC Cutoffs ##

MAD = 3

def qc_boundary(counts, k=MAD):
    x = np.log1p(counts)
    mad = median_abs_deviation(x)
    return np.exp(np.median(x) - k*mad), np.exp(np.median(x) + k*mad)

### nUMI and nGene QCs
n_counts_lower, n_counts_upper = qc_boundary(walsh_adata.obs.n_counts, k=MAD)
print('n_UMIs lower: %d upper: %d' % (n_counts_lower, n_counts_upper))

n_genes_lower, n_genes_upper = qc_boundary(walsh_adata.obs.n_genes, k=MAD)
print('n_genes lower: %d upper: %d' % (n_genes_lower, n_genes_upper))

## Plot Changes to Counts and Genes/Cell Distributions ##

fig, axs = plt.subplots(2, 2, figsize=(10, 10))  # 2x2 grid, adjust size as needed

# Plot 1: log10(n_counts)
sns.histplot(np.log10(walsh_adata.obs.n_counts), ax=axs[0, 0])
axs[0, 0].axvline(np.log10(n_counts_lower), color='red')
axs[0, 0].axvline(np.log10(n_counts_upper), color='red')
axs[0, 0].set_xlabel('log10(n_counts)', fontsize=12)
axs[0, 0].set_title("Total Counts")

# Plot 2: log10(n_genes)
sns.histplot(np.log10(walsh_adata.obs.n_genes), ax=axs[0, 1])
axs[0, 1].axvline(np.log10(n_genes_lower), color='red')
axs[0, 1].axvline(np.log10(n_genes_upper), color='red')
axs[0, 1].set_xlabel('log10(n_genes)', fontsize=12)
axs[0, 1].set_title("Number of Genes")

# Plot 3: scatter plot of n_genes vs n_counts
sns.scatterplot(x=walsh_adata.obs.n_genes, y=walsh_adata.obs.n_counts, alpha=0.5, s=0.1, ax=axs[1, 0])
axs[1, 0].axhline(n_counts_lower, color='red')
axs[1, 0].axhline(n_counts_upper, color='red')
axs[1, 0].axvline(n_genes_lower, color='red')
axs[1, 0].axvline(n_genes_upper, color='red')
axs[1, 0].set_xlabel('n_genes')
axs[1, 0].set_ylabel('n_counts')
axs[1, 0].set_title("Gene vs Count Scatter")

# Plot 4: percent_mito
sns.histplot(walsh_adata.obs.percent_mito, ax=axs[1, 1])
axs[1, 1].axvline(10, color='red')
axs[1, 1].set_xlabel('percent_mito', fontsize=12)
axs[1, 1].set_title("Percent Mitochondrial")

plt.tight_layout()
plt.show()
```

```python
## Minimum Cell Cutoffs Based on Donor-Batches ##

args_min_n_cells = 50

n_cells_before_qc = walsh_adata.obs.library.value_counts().rename_axis('library').reset_index(name='counts')
n_cells_after_qc = walsh_adata.obs[walsh_adata.obs.passed_qc].library.value_counts().rename_axis('library').reset_index(name='counts')

print('n_cells before QC:', np.sum(n_cells_before_qc[n_cells_before_qc.counts>0].counts))
print('n_cells after QC:', np.sum(n_cells_after_qc[n_cells_after_qc.counts>0].counts))

print('mean n_cells before QC', np.mean(n_cells_before_qc[n_cells_before_qc.counts>0].counts))
print('mean n_cells after QC', np.mean(n_cells_after_qc[n_cells_after_qc.counts>0].counts))

# Plot Distribution of Donor-Batches on Being Greater than Zero
plt.figure(figsize=(4, 4))
plt.title('Batch Counts')
sns.histplot(np.log10(n_cells_before_qc[n_cells_before_qc.counts>0].counts))
plt.axvline(np.log10(args_min_n_cells), color='red')
plt.xlabel('log10(n_cells)', fontsize=12)
plt.show()

### n_cells QC

#identify donor-batches with less than 50 cells of contribution
n_cells_outlier = list(n_cells_after_qc[n_cells_after_qc.counts<args_min_n_cells].library)
# set these to have failed QC as well
walsh_adata.obs.loc[walsh_adata.obs.library.isin(n_cells_outlier),'passed_qc'] = False
print('remove {0} batches that have cells less than {1}'.format(len(n_cells_outlier), args_min_n_cells))
```

```python
## Trialing QC Cutoffs Before and After ##

## Boolean Cutoffs for UMIs and Genes/Cell ##

counts_filter = (n_counts_lower < walsh_adata.obs.n_genes) & (walsh_adata.obs.n_genes < n_counts_upper)
genes_filter = (n_genes_lower < walsh_adata.obs.n_genes) & (walsh_adata.obs.n_genes < n_genes_upper)

# Plot Changes With Violins 

fig, axes = plt.subplots(2, 2, figsize=(10, 10))
axes = axes.flatten()
sns.violinplot(data=walsh_adata.obs, y='n_genes', x='Donor', ax=axes[0])
axes[0].set_title('Pre QC Genes/Cell')
plt.setp(axes[0].get_xticklabels(), rotation=30, ha='right')

sns.violinplot(data=walsh_adata.obs[counts_filter & genes_filter], x='Donor', y='n_genes', ax=axes[1])
axes[1].set_title('Post QC Genes/Cell')
plt.setp(axes[1].get_xticklabels(), rotation=30, ha='right')

sns.violinplot(data=walsh_adata.obs, y='n_counts', x='Donor', ax=axes[2])
axes[2].set_title('Pre QC Total Counts/Cell')
plt.setp(axes[2].get_xticklabels(), rotation=30, ha='right')

sns.violinplot(data=walsh_adata.obs[walsh_adata.obs.passed_qc & genes_filter], x='Donor', y='n_counts', ax=axes[3])
axes[3].set_title('Post QC Total Counts/Cell')
plt.setp(axes[3].get_xticklabels(), rotation=30, ha='right')

plt.tight_layout()
plt.show()
```

```python
## Remove Cells That Don't Meet Total Counts/Cell and Genes/Cell Thresholds ##

walsh_adata = MultimodalData(UnimodalData(walsh_adata, modality="rna"))

pg.qc_metrics(walsh_adata, 
              min_genes=n_genes_lower, max_genes=n_genes_upper,
              min_umis=n_counts_lower, max_umis=n_counts_upper,
              mito_prefix='MT-', percent_mito=10)

pg.filter_data(walsh_adata)
gc.collect()
walsh_adata = walsh_adata.to_anndata()
```

```python
## Filter Cells and Genes With Raw Cutoffs #

# store raw counts in layer
walsh_adata.layers["raw_counts"] = walsh_adata.X

# Note: No cells fail this cutoff affter using MAD based filtration
print((walsh_adata.obs.n_genes < 100).sum())
print((walsh_adata.var.n_cells < 20).any())

# Doing This For Completeness
sc.pp.filter_cells(walsh_adata, min_genes=100)
sc.pp.filter_genes(walsh_adata, min_cells=20)
```

```python
## Log Normalization ##

# normalize 
scales_counts = sc.pp.normalize_total(walsh_adata, layer='raw_counts', target_sum=None, inplace=False)

# log1p transform
walsh_adata.layers["log1p_norm"] = sc.pp.log1p(scales_counts["X"], copy=True)

# set X as log1p normed counts
walsh_adata.X = walsh_adata.layers["log1p_norm"]

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
p1 = sns.histplot(walsh_adata.obs["n_counts"], bins=100, kde=False, ax=axes[0])
axes[0].set_title("Total counts")
p2 = sns.histplot(walsh_adata.layers["log1p_norm"].sum(1), bins=100, kde=False, ax=axes[1])
axes[1].set_title("Shifted logarithm")
plt.show()
```

```python
walsh_adata[:,walsh_adata.var.robust_protein_coding_autosome]
```

```python
# Calculate Highly Variable Genes (not using deviance) ##

if not isinstance(walsh_adata, UnimodalData):
    walsh_adata = UnimodalData(walsh_adata)

## ROUND 0 ##

# calc score requires a float downcast for some reason
walsh_adata.X = walsh_adata.X.astype('float32')
pg.calc_signature_score(walsh_adata, 'cell_cycle_human')

# calculate HVGs on a protein_coding_autosomal only gene list
walsh_hvg = walsh_adata[:,walsh_adata.var.robust_protein_coding_autosome].copy()
walsh_hvg = walsh_hvg.to_anndata()
hvg = sc.pp.highly_variable_genes(walsh_hvg, layer='log1p_norm', n_top_genes=3000, inplace=False, subset=False)
hvg_list = walsh_hvg.var.index[hvg.highly_variable].tolist()
del walsh_hvg
gc.collect()

# apply that HVG list to your original anndata object
walsh_adata.var['highly_variable_features'] = False
walsh_adata.var['highly_variable_features'] = walsh_adata.var.index.isin(hvg_list)

## Dimensionality Reduction ##

# PCA

if not isinstance(walsh_adata, UnimodalData):
    walsh_adata = UnimodalData(walsh_adata)

pg.pca(walsh_adata, n_components=30)
sc.pl.pca_variance_ratio(walsh_adata, n_pcs=30, log=True)
sc.pl.pca_scatter(walsh_adata.to_anndata(), color="n_counts")
plt.show()

print('There is {0} variance explained by these components'.format(walsh_adata.uns['pca']['variance_ratio'].sum()*100))

# # Regress Out ##

pg.regress_out(walsh_adata, ['n_counts', 'percent_mito'])

## Harmony  

#pg.run_harmony(walsh_adata, batch=['Donor'], rep='pca_regressed', max_iter_harmony=20, n_comps=30)

# # UMAP

pg.neighbors(walsh_adata, rep='pca_regressed', use_cache=False, dist='l2', K=100, n_comps=30)
pg.leiden(walsh_adata, rep='pca_regressed', resolution=0.1, class_label='class')
pg.umap(walsh_adata, rep='pca_regressed', n_neighbors=15, rep_ncomps=30)
plt.show()

walsh_adata = walsh_adata.to_anndata()
```
