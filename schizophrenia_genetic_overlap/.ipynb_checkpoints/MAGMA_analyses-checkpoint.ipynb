```python
import gc
import glob
from pathlib import Path

import numpy as np
from scipy.stats import pearsonr, fisher_exact, spearmanr
from statsmodels.stats.multitest import multipletests

import pandas as pd
import scanpy as sc

import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
CSV_PATH = '/mnt/sdb/scz_meta_analysis_processed/dge_signatures/dreamlet_dges/disease_analyses/'
```

```python
## FDR Calculation Function ##

def fdr_per_celltype(dge_df):
    """Add FDR-adjusted p-values and significance per cell type."""
    dge_df['p.adjusted'] = None
    dge_df['significant_fdr'] = False
    
    for celltype, group in dge_df.groupby('assay'):
        rejected, pvals_corrected, _, _ = multipletests(group['p.value'], method='fdr_bh')
        dge_df.loc[group.index, 'p.adjusted'] = pvals_corrected
        dge_df.loc[group.index, 'significant_fdr'] = rejected
    
    return dge_df
```

```python
## Single Cell ##

dge_3q29 = pd.read_csv(Path(CSV_PATH) / 'dreamlet_purcell_dge_scz_results.csv')
dge_15q13 = pd.read_csv(Path(CSV_PATH) / 'dreamlet_walsh_15q13_dge_scz_results.csv')
dge_nrxn1 = pd.read_csv(Path(CSV_PATH) / 'dreamlet_nrxn1_combined.csv')
dge_22q11 = pd.read_csv(Path(CSV_PATH) / 'dreamlet_22q11_combined.csv')
dge_idiopathic = pd.read_csv(Path(CSV_PATH) / 'dreamlet_idiopathic_combined.csv')
```

```python
## Bulk ##

dge_3q29_bulk = pd.read_csv(Path(CSV_PATH) / 'dreamlet_purcell_dge_scz_bulk_results.csv')
dge_15q13_bulk  = pd.read_csv(Path(CSV_PATH) / 'dreamlet_walsh_15q13_dge_scz_bulk_results.csv')
dge_nrxn1_bulk  = pd.read_csv(Path(CSV_PATH) / 'dreamlet_nrxn1_combined_bulk.csv')
dge_22q11_bulk  = pd.read_csv(Path(CSV_PATH) / 'dreamlet_22q11_combined_bulk.csv')
dge_idiopathic_bulk  = pd.read_csv(Path(CSV_PATH) / 'dreamlet_idiopathic_combined_bulk.csv')

dge_idiopathic_bulk = fdr_per_celltype(dge_idiopathic_bulk)
dge_22q11_bulk = fdr_per_celltype(dge_22q11_bulk)
dge_nrxn1_bulk = fdr_per_celltype(dge_nrxn1_bulk)
```

```python
def clean_limma_dge(df):
    out = df.copy()

    out = out.rename(columns={'z.std':'DGE_STAT', 'logFC':'DGE_EFFECT', 'ID':'SYMBOL'})

    return out[['SYMBOL','assay','DGE_EFFECT','DGE_STAT','P.Value','adj.P.Val','dataset']]

def clean_dreamlet(df):

    out = df.copy()

    out = out.rename(columns={"ID":"SYMBOL", "estimate":"DGE_EFFECT",  "statistic":"DGE_STAT", "p.value":"P_VALUE"})

    out["FDR"] = np.nan

    return out[["SYMBOL", "assay", "DGE_EFFECT",  "DGE_STAT", "P_VALUE", "FDR", "n.studies", "method"]]
```

```python
dge_3q29_clean = clean_limma_dge(dge_3q29)
dge_15q13_clean = clean_limma_dge(dge_15q13)
dge_idiopathic_clean = clean_dreamlet(dge_idiopathic)
dge_22q11_clean = clean_dreamlet(dge_22q11)
dge_nrxn1_clean = clean_dreamlet(dge_nrxn1)
```

```python
scz_pgc3 = pd.read_csv('~/programs/MAGMA/results/SCZ_PGC3_gene_results_annotated.csv')

# FDR correction
rejected, p_adj, _, _ = multipletests(scz_pgc3["P"], method="fdr_bh")

scz_pgc3["p.adjusted"] = p_adj
scz_pgc3["significant_fdr"] = rejected
```

```python
scz = scz_pgc3[['SYMBOL','ZSTAT','P']].copy()

scz = scz.rename(columns={'ZSTAT':'SCZ_MAGMA_Z', 'P':'SCZ_MAGMA_P'})
```

```python
dge_idiopathic_clean_merged = dge_idiopathic_clean.merge(scz, on="SYMBOL", how="inner")
```

```python
dge_idiopathic_clean_merged.groupby("assay").size()
```

```python
from scipy.stats import fisher_exact

# Genes tested in both datasets
common_genes = set(dge_idiopathic_bulk["ID"]) & set(scz_pgc3["SYMBOL"])

# Significant genes
idio_sig = (
    set(dge_idiopathic_bulk.loc[dge_idiopathic_bulk["significant_fdr"], "ID"])
    & common_genes
)

pgc_sig = (
    set(scz_pgc3.loc[scz_pgc3["significant_fdr"], "SYMBOL"])
    & common_genes
)

# Contingency table
a = len(idio_sig & pgc_sig)                 # significant in both
b = len(idio_sig - pgc_sig)                 # idiopathic only
c = len(pgc_sig - idio_sig)                 # PGC only
d = len(common_genes - idio_sig - pgc_sig)  # neither

table = [[a, b],
         [c, d]]

odds_ratio, p = fisher_exact(table, alternative="greater")

print("Contingency table:")
print(table)
print(f"Odds ratio = {odds_ratio:.3f}")
print(f"Fisher p-value = {p:.3e}")

print(f"\nGenes tested in both: {len(common_genes)}")
print(f"Idiopathic significant: {len(idio_sig)}")
print(f"PGC significant: {len(pgc_sig)}")
print(f"Overlap: {a}")
```

```python
from scipy.stats import pearsonr

# Merge on gene symbol
merged = dge_idiopathic_bulk.merge(
    scz_pgc3,
    left_on="ID",
    right_on="SYMBOL",
    how="inner"
)

# Remove missing values just in case
merged = merged[
    merged["estimate"].notna() &
    merged["ZSTAT"].notna()
]

# Pearson correlation
r, p = pearsonr(merged["estimate"], merged["ZSTAT"])

print(f"Genes: {len(merged)}")
print(f"Pearson r = {r:.3f}")
print(f"p-value = {p:.3e}")
```

```python
## Correlate Meta-Analytic t statistic with MAGMA z stat #

cor_results = []

for assay, df in dge_idiopathic_clean_merged.groupby("assay"):

    # if len(df) < 100:
    #     continue

    rho, p = spearmanr(df["DGE_STAT"], df["SCZ_MAGMA_Z"])

    cor_results.append({"assay": assay, "n_genes": len(df),  "rho": rho, "p": p}    )

cor_results = pd.DataFrame(cor_results)

cor_results["FDR"] = multipletests(cor_results["p"], method="fdr_bh")[1]

cor_results.sort_values("FDR")
```

```python
# Check overlap between between DGE test statistic with MAGMA z stat #

for name, df in {
    "idiopathic": dge_idiopathic_clean,
    "15q13": dge_15q13_clean,
    "3q29": dge_3q29_clean
}.items():

    overlap = set(df["SYMBOL"]) & set(scz_pgc3["SYMBOL"])

    print(name, len(overlap), "overlapping genes")
```

```python
from scipy.stats import mannwhitneyu

scz_genes = set(
    scz_pgc3.loc[
        scz_pgc3["significant_fdr"],
        "SYMBOL"
    ]
)

merged["is_scz"] = merged["ID"].isin(scz_genes)

mannwhitneyu(
    merged.loc[merged.is_scz, "statistic"],
    merged.loc[~merged.is_scz, "statistic"],
    alternative="two-sided"
)
```

```python
## Top 5% MAGMA hits against top 5% of my DGEs by celltype

# background universe
background = set(scz_pgc3["SYMBOL"])

# define thresholds
top_dge = set(
    dge_idiopathic_clean
    .query("DGE_STAT > 1.96")["SYMBOL"]
)

top_scz = set(
    scz_pgc3
    .query("P < 0.05")["SYMBOL"]
)

a = len(top_dge & top_scz)
b = len(top_dge - top_scz)
c = len(top_scz - top_dge)
d = len(background - top_dge - top_scz)

table = [[a,b],
         [c,d]]

oddsratio, p = fisher_exact(table)

print(a, oddsratio, p)
```
