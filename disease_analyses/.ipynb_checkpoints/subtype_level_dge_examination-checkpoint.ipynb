```python
import os
import gc

import numpy as np
from statsmodels.stats.multitest import multipletests
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

import anndata as ad
import scanpy as sc

%matplotlib inline
```

```python
## Import In Disease Subtype Level Stats ##

CSV_PATH = '/mnt/sdb/scz_meta_analysis_processed/dge_signatures/dreamlet_dges/disease_analyses/'

dge_22q11 = pd.read_csv(CSV_PATH + "dreamlet_22q11_combined_subtype.csv")
dge_NRXN1 = pd.read_csv(CSV_PATH + "dreamlet_nrxn1_combined_subtype.csv")
dge_idiopathic = pd.read_csv(CSV_PATH + "dreamlet_idiopathic_combined_subtype.csv")

dge_3q29 = pd.read_csv(CSV_PATH + "dreamlet_purcell_dge_scz_results_subtype.csv")
dge_15q13 = pd.read_csv(CSV_PATH + "dreamlet_walsh_15q13_dge_scz_results_subtype.csv")
```

```python
## FDR Calculation Function ##

def fdr_per_celltype(dge_df):
    """Add FDR-adjusted p-values and significance per cell type."""
    dge_df['p.adjusted'] = None
    dge_df['significant_fdr'] = False
    
    for celltype, group in dge_df.groupby('assay'):
        rejected, pvals_corrected, _, _ = multipletests(group['p.value'], method='fdr_bh')
        dge_df.loc[group.index, 'p.adjusted'] = pvals_corrected
        dge_df.loc[group.index, 'significant_fdr'] = rejected
    
    return dge_df
```

```python
## FDR Correct Meta-Analyses #

dge_22q11 = fdr_per_celltype(dge_22q11)
dge_NRXN1 = fdr_per_celltype(dge_NRXN1)
dge_idiopathic = fdr_per_celltype(dge_idiopathic)
```

```python
# Filter On Significance #

dge_22q11['significant'] = dge_22q11['significant_fdr']
dge_NRXN1['significant'] = dge_NRXN1['significant_fdr']
dge_idiopathic['significant'] = dge_idiopathic['significant_fdr']
dge_3q29['significant'] = (dge_3q29['adj.P.Val'] < 0.05)
dge_15q13['significant'] = (dge_15q13['adj.P.Val'] < 0.05)
```

```python
dge_3q29['estimate'] = dge_3q29['logFC'] 
dge_15q13['estimate'] = dge_15q13['logFC']
```

```python
dge = pd.concat([
                dge_22q11.assign(group="22q11"),
                dge_NRXN1.assign(group="NRXN1"),
                dge_3q29.assign(group="3q29"),
                dge_15q13.assign(group="15q13"),
                dge_idiopathic.assign(group="Idiopathic")],
    ignore_index=True)
```

```python
dge[(dge['significant'] & np.abs(dge['estimate'] > 1))]['assay'].value_counts()
```

```python

fig, ax = plt.subplots(figsize=(12, 6))

sns.barplot(
    data=dge[dge["significant"]],
    x="group",
    y="estimate",
    hue="assay",
    ax=ax,
)

ax.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    borderaxespad=0,
)

plt.tight_layout()
```

```python
for group in dge['group'].unique():
    temp_dge = dge[dge['group'] == group]

    fig, ax = plt.subplots(figsize=(12, 8))
    plt.title(group)
    sns.violinplot(
    data=temp_dge[temp_dge["significant"]],
    x="group",
    y="estimate",
    hue="assay",
    ax=ax)
    # ax.legend_.remove()
    ax.legend(
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    borderaxespad=0,
)
    plt.tight_layout()
    plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 8))

sns.violinplot(
    data=dge[dge["significant"]],
    x="group",
    y="estimate",
    hue="assay",
    ax=ax,
)

ax.legend_.remove()

plt.tight_layout()
```
